```python
import os
import pandas as pd
from Bio import SeqIO
from collections import defaultdict
```

```python
# Root directory
root_dir = "./"

# data[gene][curation_level] = count
data = defaultdict(lambda: defaultdict(int))

# Loop through each curation directory
for curation_level in os.listdir(root_dir):
    level_path = os.path.join(root_dir, curation_level)

    if not os.path.isdir(level_path):
        continue

    print(f"\nScanning: {curation_level}")

    for file in os.listdir(level_path):
        if file.lower().endswith((".fasta", ".fa", ".faa")):

            fasta_path = os.path.join(level_path, file)

            # -----------------------------
            # Extract gene from filename
            # -----------------------------
            name = os.path.splitext(file)[0]
            gene = name.split("_")[0]

            # -----------------------------
            # Count sequences
            # -----------------------------
            count = sum(1 for _ in SeqIO.parse(fasta_path, "fasta"))

            data[gene][curation_level] += count

            print(f"  {file} -> {gene} | {curation_level} | {count}")

# -----------------------------
# Build dataframe
# -----------------------------
df = pd.DataFrame(data).T.fillna(0).astype(int)

# Order rows and columns
df = df.sort_index()
df = df.reindex(sorted(df.columns), axis=1)

# Add total row
df.loc["TOTAL"] = df.sum()

# Name index column
df.index.name = "gene"

# Save summary
df.to_csv("final_fastas_summary.csv")

print("\n✅ Summary table:")
print(df)
```
